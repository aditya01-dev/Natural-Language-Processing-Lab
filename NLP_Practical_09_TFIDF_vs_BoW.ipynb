```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import pandas as pd

print("Libraries imported successfully.")
```

```python
documents = [
    "Natural language processing is interesting.",
    "Natural language processing is useful.",
    "Machine learning is useful for natural language processing."
]

print("Documents:\n")

for i, document in enumerate(documents, 1):
    print(f"Document {i}: {document}")
```

```python
bow_vectorizer = CountVectorizer()

bow_matrix = bow_vectorizer.fit_transform(documents)

bow_vocabulary = bow_vectorizer.get_feature_names_out()

print("Bag-of-Words Vocabulary:")
print(bow_vocabulary)

print("\nBag-of-Words Matrix:")
print(bow_matrix.toarray())
```

```python
bow_table = pd.DataFrame(
    bow_matrix.toarray(),
    columns=bow_vocabulary
)

bow_table.index = [
    "Document 1",
    "Document 2",
    "Document 3"
]

print("Bag-of-Words Representation:")
display(bow_table)
```

```python
tfidf_vectorizer = TfidfVectorizer()

tfidf_matrix = tfidf_vectorizer.fit_transform(documents)

tfidf_vocabulary = tfidf_vectorizer.get_feature_names_out()

print("TF-IDF Vocabulary:")
print(tfidf_vocabulary)

print("\nTF-IDF Matrix:")
print(tfidf_matrix.toarray())
```

```python
tfidf_table = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf_vocabulary
)

tfidf_table.index = [
    "Document 1",
    "Document 2",
    "Document 3"
]

print("TF-IDF Representation:")
display(tfidf_table.round(3))
```

```python
print("Comparison of BoW and TF-IDF")
print("=" * 60)

print("\nBag-of-Words:")
display(bow_table)

print("\nTF-IDF:")
display(tfidf_table.round(3))
```

```python
word = "language"

word_index_bow = list(bow_vocabulary).index(word)
word_index_tfidf = list(tfidf_vocabulary).index(word)

print(f"Word: {word}")
print("-" * 40)

print("BoW values:")
print(bow_matrix.toarray()[:, word_index_bow])

print("\nTF-IDF values:")
print(tfidf_matrix.toarray()[:, word_index_tfidf])
```

```python
for i, row in enumerate(tfidf_matrix.toarray(), 1):
    max_index = row.argmax()
    max_word = tfidf_vocabulary[max_index]
    max_value = row[max_index]

    print(
        f"Document {i}: "
        f"{max_word} -> {max_value:.3f}"
    )
```

```python
print("Main Difference:")
print("-" * 50)

print("BoW counts how many times a word occurs in a document.")
print("TF-IDF assigns higher importance to words that are frequent")
print("in a document but relatively less common across the corpus.")
```

```python
# TF-IDF Concept:
# TF  = Term Frequency
# IDF = Inverse Document Frequency
# TF-IDF = TF × IDF
# A word that appears frequently in one document but appears in fewer
# documents across the corpus generally receives a higher TF-IDF weight.
```

```python
# Result:
# The given corpus was successfully converted into both Bag-of-Words
# and TF-IDF representations.
#
# The comparison showed that BoW represents documents using raw word
# frequencies, whereas TF-IDF assigns weighted values based on the
# importance of terms within the corpus.
#
# Conclusion:
# TF-IDF improves upon simple word-frequency representation by reducing
# the importance of terms that occur frequently across many documents
# and emphasizing more informative terms.
```
