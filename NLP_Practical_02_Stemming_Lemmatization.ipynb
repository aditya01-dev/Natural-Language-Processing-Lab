```python
import nltk
import spacy

from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
!python -m spacy download en_core_web_sm
```

```python
nlp = spacy.load("en_core_web_sm")

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
```

```python
text = """
The students are studying natural language processing.
They studied different techniques and are learning continuously.
The studies are interesting and useful.
The researchers were analyzing the results carefully.
"""

print("Original Text:")
print(text)
```

```python
words = word_tokenize(text)

print("Tokenized Words:")
print(words)
```

```python
stemmed_words = [stemmer.stem(word) for word in words]

print("Stemmed Words:")
print(stemmed_words)
```

```python
lemmatized_words = [
    lemmatizer.lemmatize(word)
    for word in words
]

print("Lemmatized Words using NLTK:")
print(lemmatized_words)
```

```python
doc = nlp(text)

print("Lemmatization using spaCy:\n")

for token in doc:
    print(f"{token.text:15} -> {token.lemma_}")
```

```python
comparison_words = [
    "studying",
    "studied",
    "studies",
    "learning",
    "researchers",
    "analyzing"
]

print(f"{'Word':15} {'Stem':15} {'Lemma':15}")
print("-" * 45)

for word in comparison_words:
    stem = stemmer.stem(word)
    lemma = lemmatizer.lemmatize(word)
    print(f"{word:15} {stem:15} {lemma:15}")
```

```python
print(f"{'Word':15} {'POS':10} {'Lemma':15}")
print("-" * 40)

for token in doc:
    if token.is_alpha:
        print(f"{token.text:15} {token.pos_:10} {token.lemma_:15}")
```

```python
## Result

#Stemming and lemmatization were successfully performed on the given text.

#- Stemming was performed using NLTK's Porter Stemmer.
#- Lemmatization was performed using NLTK's WordNet Lemmatizer.
#- Lemmatization was also demonstrated using spaCy.
#- The results showed that stemming generally produces truncated root-like forms, while lemmatization attempts to produce meaningful dictionary base forms.
```

```python
## Conclusion

#Stemming and lemmatization are important Natural Language Processing preprocessing techniques. Stemming reduces words to a root-like form by removing word endings, whereas lemmatization converts words into their meaningful base or dictionary forms.
```
