```python
import nltk
import spacy

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
!python -m spacy download en_core_web_sm
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
text = """
Natural Language Processing is a field of Artificial Intelligence.
It is used to process and understand human language.
The field has many practical applications in modern technology.
Students can learn NLP to build intelligent and useful applications.
"""

print("Original Document:")
print(text)
```

```python
words = word_tokenize(text)

print("Tokenized Words:")
print(words)
```

```python
stop_words = set(stopwords.words('english'))

print("Total Number of NLTK English Stop Words:", len(stop_words))

print("\nSample Stop Words:")
print(list(stop_words)[:30])
```

```python
filtered_words_nltk = [
    word
    for word in words
    if word.lower() not in stop_words
]

print("Words after Stop-word Removal using NLTK:")
print(filtered_words_nltk)
```

```python
filtered_words_nltk_clean = [
    word
    for word in words
    if word.lower() not in stop_words and word.isalpha()
]

print("Cleaned Words after Stop-word and Punctuation Removal:")
print(filtered_words_nltk_clean)
```

```python
cleaned_text_nltk = " ".join(filtered_words_nltk_clean)

print("Final Document after NLTK Stop-word Removal:")
print(cleaned_text_nltk)
```

```python
doc = nlp(text)

filtered_words_spacy = [
    token.text
    for token in doc
    if not token.is_stop and not token.is_punct
]

print("Words after Stop-word Removal using spaCy:")
print(filtered_words_spacy)
```

```python
cleaned_text_spacy = " ".join(filtered_words_spacy)

print("Final Document after spaCy Stop-word Removal:")
print(cleaned_text_spacy)
```

```python
original_word_count = len([
    word for word in words
    if word.isalpha()
])

nltk_word_count = len(filtered_words_nltk_clean)

spacy_word_count = len([
    token
    for token in doc
    if not token.is_stop
    and not token.is_punct
    and token.is_alpha
])

print("Word Count Comparison")
print("-" * 40)

print(f"Original word count : {original_word_count}")
print(f"After NLTK         : {nltk_word_count}")
print(f"After spaCy        : {spacy_word_count}")
```

```python
removed_words = [
    word
    for word in words
    if word.lower() in stop_words
]

print("Stop Words Removed by NLTK:")
print(removed_words)
```

```python
# Result:
# Stop words were successfully identified and removed from the given document
# using both NLTK and spaCy.
#
# Conclusion:
# Stop-word removal is an important preprocessing technique in Natural
# Language Processing. It helps reduce frequently occurring words that
# may contribute less useful information for certain NLP applications.
```
