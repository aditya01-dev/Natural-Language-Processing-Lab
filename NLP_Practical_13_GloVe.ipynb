```python
import gensim.downloader as api
import numpy as np

print("Libraries imported successfully.")
```

```python
print("Loading pre-trained GloVe model...")

glove_model = api.load("glove-wiki-gigaword-50")

print("GloVe model loaded successfully.")
```

```python
print("GloVe Model Information")
print("-" * 40)

print("Vocabulary Size:", len(glove_model))
print("Vector Dimension:", glove_model.vector_size)
```

```python
word = "language"

if word in glove_model:
    print(f"'{word}' is present in the GloVe vocabulary.")
else:
    print(f"'{word}' is not present in the GloVe vocabulary.")
```

```python
word = "language"

vector = glove_model[word]

print(f"Word: {word}")
print("\nGloVe Vector:")
print(vector)

print("\nVector Dimension:", len(vector))
```

```python
word = "language"

print(f"First 10 values of the GloVe vector for '{word}':")
print(glove_model[word][:10])
```

```python
words = ["language", "computer", "machine", "learning", "artificial"]

print("Word Vector Dimensions:\n")

for word in words:
    if word in glove_model:
        print(f"{word:15} -> {glove_model[word].shape}")
    else:
        print(f"{word:15} -> Word not found")
```

```python
word = "language"

print(f"Words most similar to '{word}':\n")

similar_words = glove_model.most_similar(word, topn=10)

for similar_word, score in similar_words:
    print(f"{similar_word:15} -> {score:.4f}")
```

```python
word1 = "language"
word2 = "linguistic"

similarity = glove_model.similarity(word1, word2)

print(f"Word 1: {word1}")
print(f"Word 2: {word2}")
print(f"\nCosine Similarity: {similarity:.4f}")
```

```python
word_pairs = [
    ("language", "linguistic"),
    ("computer", "technology"),
    ("king", "queen"),
    ("apple", "banana")
]

print("Word Similarity Results")
print("-" * 50)

for word1, word2 in word_pairs:
    if word1 in glove_model and word2 in glove_model:
        score = glove_model.similarity(word1, word2)
        print(f"{word1:12} <-> {word2:12} : {score:.4f}")
    else:
        print(f"{word1} or {word2} not found in vocabulary")
```

```python
selected_words = ["king", "queen", "man", "woman"]

vectors = glove_model[[word for word in selected_words if word in glove_model]]

print("Shape of selected word vectors:")
print(vectors.shape)

print("\nEach row represents one word vector.")
```

```python
result = glove_model.most_similar(
    positive=["king", "woman"],
    negative=["man"],
    topn=5
)

print("Vector Arithmetic Result:")
print("\nking + woman - man\n")

for word, score in result:
    print(f"{word:15} -> {score:.4f}")
```

```python
word = "language"

vector = glove_model[word]

norm = np.linalg.norm(vector)

print(f"Word: {word}")
print(f"Vector Norm: {norm:.4f}")
```

```python
import pandas as pd

selected_words = ["language", "computer", "learning"]

data = []

for word in selected_words:
    if word in glove_model:
        data.append([word] + list(glove_model[word][:10]))

columns = ["Word"] + [f"Dim_{i+1}" for i in range(10)]

vector_table = pd.DataFrame(data, columns=columns)

print("First 10 Dimensions of Selected GloVe Vectors:")
display(vector_table)
```
