```python
# Name: Aditya kumar Singh
# Practical 31: Document Classification and Separation
# Objective: Classify documents into categories and separate them accordingly.

# Import required libraries
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline

# Sample documents
documents = [
    "The football team won the championship match.",
    "The player scored a goal in the final game.",
    "Cricket is a popular sport in India.",
    "Artificial intelligence is transforming technology.",
    "Machine learning helps computers learn from data.",
    "Python is widely used for software development.",
    "The government introduced a new policy.",
    "The election campaign started yesterday.",
    "Parliament discussed the national budget.",
    "Students are preparing for university examinations.",
    "Teachers use digital tools for learning.",
    "Education improves knowledge and career opportunities."
]

# Document labels
labels = [
    "Sports", "Sports", "Sports",
    "Technology", "Technology", "Technology",
    "Politics", "Politics", "Politics",
    "Education", "Education", "Education"
]

# Create a TF-IDF and Naive Bayes classification pipeline
classifier = Pipeline([
    ("tfidf", TfidfVectorizer()),
    ("model", MultinomialNB())
])

# Train the classifier
classifier.fit(documents, labels)

# New documents to classify
new_documents = [
    "The team played an excellent match and scored goals.",
    "Python programming is useful for artificial intelligence.",
    "The minister announced a new government policy.",
    "Students attended classes and completed their assignments.",
    "The cricket player won the tournament."
]

# Predict document categories
predictions = classifier.predict(new_documents)

# Separate documents according to category
separated_documents = {}

for document, category in zip(new_documents, predictions):
    separated_documents.setdefault(category, []).append(document)

# Display classification results
print("Document Classification Results:\n")

for document, category in zip(new_documents, predictions):
    print(f"Document: {document}")
    print(f"Category: {category}\n")

# Display separated documents
print("=" * 50)
print("DOCUMENTS SEPARATED BY CATEGORY")
print("=" * 50)

for category, docs in separated_documents.items():
    print(f"\n{category}:")
    for doc in docs:
        print("-", doc)
```
