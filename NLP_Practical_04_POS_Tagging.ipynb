```python
import nltk
import spacy

from nltk.tokenize import word_tokenize
from nltk import pos_tag
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
!python -m spacy download en_core_web_sm
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
sentence = "The intelligent student is developing a natural language processing project."

print("Input Sentence:")
print(sentence)
```

```python
words = word_tokenize(sentence)

print("Tokenized Words:")
print(words)
```

```python
nltk_tags = pos_tag(words)

print("POS Tags using NLTK:\n")

for word, tag in nltk_tags:
    print(f"{word:15} -> {tag}")
```

```python
# Download the NLTK tagset resource
nltk.download('tagsets_json')

from nltk.help import upenn_tagset

print("Meaning of selected POS tags:\n")

selected_tags = sorted(set(tag for _, tag in nltk_tags))

for tag in selected_tags:
    print(f"{tag}:")
    upenn_tagset(tag)
    print()
```

```python
doc = nlp(sentence)

print("POS Tags using spaCy:\n")

for token in doc:
    print(
        f"{token.text:15} -> "
        f"{token.pos_:10} -> "
        f"{token.tag_}"
    )
```

```python
print(f"{'Word':15} {'POS':12} {'Detailed Tag':15}")
print("-" * 45)

for token in doc:
    print(f"{token.text:15} {token.pos_:12} {token.tag_:15}")
```

```python
print(f"{'Word':15} {'NLTK Tag':15} {'spaCy POS':15} {'spaCy Tag':15}")
print("-" * 65)

for i, token in enumerate(doc):
    nltk_word, nltk_tag = nltk_tags[i]

    print(
        f"{nltk_word:15} "
        f"{nltk_tag:15} "
        f"{token.pos_:15} "
        f"{token.tag_:15}"
    )
```

```python
nouns = [token.text for token in doc if token.pos_ == "NOUN"]
verbs = [token.text for token in doc if token.pos_ == "VERB"]
adjectives = [token.text for token in doc if token.pos_ == "ADJ"]
determiners = [token.text for token in doc if token.pos_ == "DET"]

print("Nouns:", nouns)
print("Verbs:", verbs)
print("Adjectives:", adjectives)
print("Determiners:", determiners)
```

```python
# Result:
# Part-of-Speech tagging was successfully performed on the given sentence
# using both NLTK and spaCy.
#
# Conclusion:
# POS tagging assigns grammatical categories to words and provides
# syntactic information about a sentence. It is an important NLP
# preprocessing and linguistic analysis technique used in tasks such
# as parsing, chunking, information extraction, and text understanding.
```
