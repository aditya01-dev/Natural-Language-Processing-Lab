```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Libraries imported successfully.")
```

```python
texts = [
    "Artificial intelligence is transforming modern technology.",
    "Machine learning algorithms are used to analyze large datasets.",
    "Deep learning models are used in computer vision applications.",
    "Natural language processing helps computers understand human language.",
    "Python is widely used for machine learning and data science.",
    "Cloud computing provides scalable computing resources.",
    "Software engineers develop applications using programming languages.",
    "Data science combines statistics programming and machine learning.",

    "Football is one of the most popular sports in the world.",
    "The cricket team won the match after scoring many runs.",
    "The player scored a goal in the final minute of the game.",
    "Basketball players practice shooting and defensive techniques.",
    "The tennis tournament attracted many professional players.",
    "The team trained hard before the championship match.",
    "The athlete won a gold medal in the competition.",
    "Cricket players improve their batting and bowling skills."
]

labels = [
    "Technology",
    "Technology",
    "Technology",
    "Technology",
    "Technology",
    "Technology",
    "Technology",
    "Technology",

    "Sports",
    "Sports",
    "Sports",
    "Sports",
    "Sports",
    "Sports",
    "Sports",
    "Sports"
]

data = pd.DataFrame({
    "Text": texts,
    "Category": labels
})

print("Dataset:")
display(data)
```

```python
print("Class Distribution:")
print(data["Category"].value_counts())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    data["Text"],
    data["Category"],
    test_size=0.25,
    random_state=42,
    stratify=data["Category"]
)

print("Training samples:", len(X_train))
print("Testing samples :", len(X_test))
```

```python
tfidf_vectorizer = TfidfVectorizer()

X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)
X_test_tfidf = tfidf_vectorizer.transform(X_test)

print("TF-IDF vectorization completed.")
print("Training matrix shape:", X_train_tfidf.shape)
print("Testing matrix shape :", X_test_tfidf.shape)
```

```python
feature_names = tfidf_vectorizer.get_feature_names_out()

print("TF-IDF Vocabulary:\n")
print(feature_names)
```

```python
tfidf_table = pd.DataFrame(
    X_train_tfidf.toarray(),
    columns=feature_names
)

print("First Training Document - TF-IDF Representation:")
display(tfidf_table.head(1).round(3))
```

```python
nb_model = MultinomialNB()

nb_model.fit(X_train_tfidf, y_train)

print("Multinomial Naïve Bayes model trained successfully.")
```

```python
nb_predictions = nb_model.predict(X_test_tfidf)

print("Naïve Bayes Predictions:")
print(nb_predictions)
```

```python
nb_accuracy = accuracy_score(y_test, nb_predictions)

print("Naïve Bayes Accuracy:")
print(f"{nb_accuracy:.4f}")

print("\nClassification Report:")
print(classification_report(y_test, nb_predictions))
```

```python
svm_model = LinearSVC(random_state=42)

svm_model.fit(X_train_tfidf, y_train)

print("Linear SVM model trained successfully.")
```

```python
svm_model = LinearSVC(random_state=42)

svm_model.fit(X_train_tfidf, y_train)

print("Linear SVM model trained successfully.")
```

```python
svm_predictions = svm_model.predict(X_test_tfidf)

print("SVM Predictions:")
print(svm_predictions)
```

```python
svm_accuracy = accuracy_score(y_test, svm_predictions)

print("SVM Accuracy:")
print(f"{svm_accuracy:.4f}")

print("\nClassification Report:")
print(classification_report(y_test, svm_predictions))
```

```python
comparison = pd.DataFrame({
    "Classifier": [
        "Multinomial Naïve Bayes",
        "Linear SVM"
    ],
    "Accuracy": [
        nb_accuracy,
        svm_accuracy
    ]
})

print("Classifier Comparison:")
display(comparison.round(4))
```

```python
nb_cm = confusion_matrix(y_test, nb_predictions)
svm_cm = confusion_matrix(y_test, svm_predictions)

print("Naïve Bayes Confusion Matrix:")
print(nb_cm)

print("\nLinear SVM Confusion Matrix:")
print(svm_cm)
```

```python
new_text = [
    "The software uses machine learning algorithms to process data.",
    "The football player scored two goals in the match."
]

new_text_tfidf = tfidf_vectorizer.transform(new_text)

nb_new_predictions = nb_model.predict(new_text_tfidf)

print("New Text Classification using Naïve Bayes:\n")

for text, prediction in zip(new_text, nb_new_predictions):
    print(f"Text: {text}")
    print(f"Predicted Category: {prediction}\n")
```

```python
svm_new_predictions = svm_model.predict(new_text_tfidf)

print("New Text Classification using SVM:\n")

for text, prediction in zip(new_text, svm_new_predictions):
    print(f"Text: {text}")
    print(f"Predicted Category: {prediction}\n")
```

```python
prediction_table = pd.DataFrame({
    "Text": new_text,
    "Naïve Bayes": nb_new_predictions,
    "SVM": svm_new_predictions
})

display(prediction_table)
```

```python
print("Model Summary")
print("-" * 45)

print(f"Training Samples : {len(X_train)}")
print(f"Testing Samples  : {len(X_test)}")
print(f"Vocabulary Size  : {len(feature_names)}")

print(f"\nNaïve Bayes Accuracy : {nb_accuracy:.4f}")
print(f"SVM Accuracy         : {svm_accuracy:.4f}")
```
