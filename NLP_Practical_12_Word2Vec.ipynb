```python
from gensim.models import Word2Vec

print("Gensim imported successfully.")
```

```python
corpus = [
    ["natural", "language", "processing", "is", "important"],
    ["natural", "language", "processing", "uses", "machine", "learning"],
    ["machine", "learning", "helps", "computers", "understand", "language"],
    ["deep", "learning", "is", "used", "for", "natural", "language", "processing"],
    ["natural", "language", "models", "learn", "from", "text", "data"],
    ["word", "embeddings", "represent", "words", "as", "vectors"],
    ["word2vec", "learns", "useful", "word", "embeddings"],
    ["machine", "learning", "and", "deep", "learning", "are", "used", "in", "nlp"]
]

print("Custom Corpus:\n")

for i, sentence in enumerate(corpus, 1):
    print(f"{i}. {' '.join(sentence)}")
```

```python
cbow_model = Word2Vec(
    sentences=corpus,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4,
    sg=0,
    epochs=100
)

print("CBOW Word2Vec model trained successfully.")
```

```python
vocabulary = list(cbow_model.wv.key_to_index.keys())

print("Vocabulary:")
print(vocabulary)
```

```python
word = "language"

vector = cbow_model.wv[word]

print(f"Word: {word}")
print("\nWord Vector:")
print(vector)
print(f"\nVector Dimension: {len(vector)}")
```

```python
print("First 10 values of the word vector:")
print(cbow_model.wv["language"][:10])
```

```python
word = "language"

print(f"Words similar to '{word}':\n")

similar_words = cbow_model.wv.most_similar(word, topn=5)

for similar_word, similarity in similar_words:
    print(f"{similar_word:15} -> {similarity:.4f}")
```

```python
skipgram_model = Word2Vec(
    sentences=corpus,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4,
    sg=1,
    epochs=100
)

print("Skip-Gram Word2Vec model trained successfully.")
```

```python
print("CBOW vector dimension:")
print(len(cbow_model.wv["language"]))

print("\nSkip-Gram vector dimension:")
print(len(skipgram_model.wv["language"]))
```

```python
word = "language"

print(f"CBOW - Similar words to '{word}':")
for similar_word, similarity in cbow_model.wv.most_similar(word, topn=5):
    print(f"{similar_word:15} -> {similarity:.4f}")

print("\nSkip-Gram - Similar words:")
for similar_word, similarity in skipgram_model.wv.most_similar(word, topn=5):
    print(f"{similar_word:15} -> {similarity:.4f}")
```

```python
word1 = "language"
word2 = "processing"

similarity = cbow_model.wv.similarity(word1, word2)

print(f"Word 1: {word1}")
print(f"Word 2: {word2}")
print(f"\nCosine Similarity: {similarity:.4f}")
```

```python
# Vector Arithmetic Example

positive_words = ["language"]
negative_words = ["machine"]

result = cbow_model.wv.most_similar(
    positive=positive_words,
    negative=negative_words,
    topn=5
)

print("Vector Arithmetic Result:\n")

for word, score in result:
    print(f"{word:15} -> {score:.4f}")
```

```python
print("Word2Vec Model Information")
print("-" * 40)

print("Vocabulary Size:", len(cbow_model.wv))
print("Vector Size:", cbow_model.vector_size)
print("Window Size:", cbow_model.window)
print("Minimum Word Count:", cbow_model.min_count)
print("Training Epochs:", cbow_model.epochs)
```

```python
cbow_model.save("word2vec_cbow.model")
skipgram_model.save("word2vec_skipgram.model")

print("Both Word2Vec models saved successfully.")
```

```python
loaded_model = Word2Vec.load("word2vec_cbow.model")

print("CBOW model loaded successfully.")

print("\nVector for 'language':")
print(loaded_model.wv["language"][:10])
```
