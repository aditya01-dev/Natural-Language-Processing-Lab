```python
# ============================================================
# PRACTICAL 20: INFORMATION EXTRACTION FROM DOCUMENTS
# Course Outcome: CO2
# ============================================================
# Name: Aditya kumar Singh
# OBJECTIVE:
# To extract useful information from structured and unstructured
# documents using NLP and regular expressions.

!pip -q install spacy
!python -m spacy download en_core_web_sm -q

# Import libraries
import re
import spacy
import pandas as pd
from IPython.display import display

# Load spaCy model
nlp = spacy.load("en_core_web_sm")

# ------------------------------------------------------------
# 1. SAMPLE DOCUMENT
# ------------------------------------------------------------

document = """
Rahul Sharma joined Microsoft India in New Delhi on 15 January 2025.
He works as a Software Engineer in the Artificial Intelligence department.
His email is rahul.sharma@gmail.com and his contact number is
9876543210.

Microsoft is developing AI solutions and cloud computing services.
Rahul attended a conference in Bengaluru on 20 February 2025.
"""

print("INPUT DOCUMENT:")
print(document)

# ------------------------------------------------------------
# 2. NAMED ENTITY EXTRACTION
# ------------------------------------------------------------

doc = nlp(document)

entities = []

for ent in doc.ents:
    entities.append({
        "Entity": ent.text,
        "Entity_Type": ent.label_,
        "Description": spacy.explain(ent.label_)
    })

entity_df = pd.DataFrame(entities)

print("\nNAMED ENTITIES:")
display(entity_df)

# ------------------------------------------------------------
# 3. EMAIL EXTRACTION
# ------------------------------------------------------------

emails = re.findall(
    r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}",
    document
)

print("\nEMAIL ADDRESSES:")
print(emails)

# ------------------------------------------------------------
# 4. PHONE NUMBER EXTRACTION
# ------------------------------------------------------------

phones = re.findall(
    r"\b(?:\+91[-\s]?)?[6-9]\d{9}\b",
    document
)

print("\nPHONE NUMBERS:")
print(phones)

# ------------------------------------------------------------
# 5. DATE EXTRACTION
# ------------------------------------------------------------

dates = re.findall(
    r"\b\d{1,2}\s+[A-Za-z]+\s+\d{4}\b",
    document
)

print("\nDATES:")
print(dates)

# ------------------------------------------------------------
# 6. KEYWORD EXTRACTION
# ------------------------------------------------------------

keywords = [
    "Artificial Intelligence",
    "Software Engineer",
    "cloud computing",
    "conference"
]

extracted_keywords = [
    keyword for keyword in keywords
    if keyword.lower() in document.lower()
]

print("\nKEYWORDS:")
print(extracted_keywords)

# ------------------------------------------------------------
# 7. FINAL INFORMATION EXTRACTION REPORT
# ------------------------------------------------------------

information = {
    "Emails": emails,
    "Phone Numbers": phones,
    "Dates": dates,
    "Keywords": extracted_keywords,
    "Named Entities": [
        {"text": ent.text, "type": ent.label_}
        for ent in doc.ents
    ]
}

print("\nFINAL INFORMATION EXTRACTION REPORT:")
for key, value in information.items():
    print(f"\n{key}:")
    print(value)

# ------------------------------------------------------------
# 8. RESULT AND CONCLUSION
# ------------------------------------------------------------

print("\nRESULT:")
print("Information was successfully extracted from the document.")

print("\nCONCLUSION:")
print("NLP and regular expressions can be used to identify and")
print("extract important information from unstructured text.")
```
