```python
import re

print("Libraries imported successfully.")
```

```python
text = """
Alice went to the library because she needed a book.
The librarian helped Alice find the book, and she thanked her.
Alice took the book home because she wanted to study it.
"""

print("Original Text:")
print(text)
```

```python
sentences = [s.strip() for s in text.strip().split(".") if s.strip()]

print("Sentences:\n")

for i, sentence in enumerate(sentences, 1):
    print(f"{i}. {sentence}.")
```

```python
coreference_rules = {
    "she": "Alice",
    "her": "Alice",
    "he": "Alice",
    "him": "Alice",
    "his": "Alice",
    "it": "the book"
}

print("Coreference Rules:")
for pronoun, entity in coreference_rules.items():
    print(f"{pronoun} -> {entity}")
```

```python
resolved_text = text

for pronoun, entity in coreference_rules.items():
    pattern = r"\b" + pronoun + r"\b"
    resolved_text = re.sub(
        pattern,
        entity,
        resolved_text,
        flags=re.IGNORECASE
    )

print("Resolved Text:")
print(resolved_text)
```

```python
print("Detected Coreferences:\n")

for pronoun, entity in coreference_rules.items():
    if re.search(r"\b" + pronoun + r"\b", text, flags=re.IGNORECASE):
        print(f"{pronoun} -> {entity}")
```

```python
print("Original Text:")
print(text)

print("=" * 70)

print("Resolved Text:")
print(resolved_text)
```

```python
# Result:
# Coreference resolution was successfully demonstrated using a simple
# rule-based approach. Pronouns in the text were identified and mapped
# to their corresponding entities.
#
# Conclusion:
# Coreference resolution helps identify different expressions that refer
# to the same entity. It improves document-level understanding and is
# useful in tasks such as information extraction, summarization,
# question answering, and document analysis.
```
