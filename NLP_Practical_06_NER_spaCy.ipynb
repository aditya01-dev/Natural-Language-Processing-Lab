```python
import spacy
from spacy import displacy
```

```python
!python -m spacy download en_core_web_sm
```

```python
nlp = spacy.load("en_core_web_sm")

print("spaCy English model loaded successfully.")
```

```python
text = """
Barack Obama was born in Honolulu, Hawaii, on August 4, 1961.
He served as the 44th President of the United States from 2009 to 2017.
During his presidency, he worked with organizations such as the United Nations
and visited countries including India, Germany, and Japan.
"""

print("Input Text:")
print(text)
```

```python
doc = nlp(text)

print("Text processed successfully using spaCy.")
```

```python
print("Named Entities:\n")

for ent in doc.ents:
    print(f"Entity: {ent.text}")
    print(f"Label : {ent.label_}")
    print()
```

```python
print("Named Entity Labels:\n")

for ent in doc.ents:
    print(f"{ent.text:25} -> {ent.label_:10} -> {spacy.explain(ent.label_)}")
```

```python
print(f"{'Entity':25} {'Label':12} {'Description'}")
print("-" * 70)

for ent in doc.ents:
    description = spacy.explain(ent.label_)
    print(f"{ent.text:25} {ent.label_:12} {description}")
```

```python
from collections import Counter

entity_counts = Counter(ent.label_ for ent in doc.ents)

print("Entity Type Counts:\n")

for label, count in entity_counts.items():
    print(f"{label:10} -> {count}")
```

```python
print("Entities by Category:\n")

for label in sorted(set(ent.label_ for ent in doc.ents)):
    entities = [ent.text for ent in doc.ents if ent.label_ == label]

    print(f"{label} ({spacy.explain(label)}):")

    for entity in entities:
        print(f"  - {entity}")

    print()
```

```python
displacy.render(doc, style="ent", jupyter=True)
```

```python
print(f"{'Token':20} {'Entity IOB':15} {'Entity Type'}")
print("-" * 55)

for token in doc:
    print(
        f"{token.text:20} "
        f"{token.ent_iob_:15} "
        f"{token.ent_type_}"
    )
```
