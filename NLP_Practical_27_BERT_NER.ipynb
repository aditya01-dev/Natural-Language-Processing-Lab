```python
# Name: Aditya kumar Singh
# Practical 27: BERT Named Entity Recognition
# Objective: Identify named entities in text using a pretrained BERT model.

# Install required library
!pip -q install transformers

# Import required libraries
from transformers import pipeline

# Load pretrained BERT NER model
ner = pipeline(
    "ner",
    model="dslim/bert-base-NER",
    aggregation_strategy="simple"
)

# Input text
text = """
Araib Khan is studying Artificial Intelligence at NIET in Greater Noida.
Google is headquartered in California and OpenAI is based in San Francisco.
"""

# Perform Named Entity Recognition
entities = ner(text)

# Display the input text
print("Input Text:")
print(text)

# Display extracted entities
print("\nNamed Entities:")
for entity in entities:
    print(
        f"Entity: {entity['word']}, "
        f"Label: {entity['entity_group']}, "
        f"Confidence: {entity['score']:.4f}"
    )
```
