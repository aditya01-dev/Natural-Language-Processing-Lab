```python


# Step 1: Install required libraries

!pip install -q textblob nltk pandas matplotlib

# Step 2: Import libraries

import nltk
import pandas as pd
import matplotlib.pyplot as plt

from textblob import TextBlob
from nltk.sentiment.vader import SentimentIntensityAnalyzer

print("Libraries imported successfully.")

# Step 3: Download VADER lexicon

nltk.download('vader_lexicon', quiet=True)

# Step 4: Initialize VADER

vader_analyzer = SentimentIntensityAnalyzer()

print("VADER initialized successfully.")

# Step 5: Create a sample review dataset

reviews = [
    "The product is excellent and works perfectly.",
    "I absolutely love this phone. Its battery is amazing!",
    "The customer service was terrible and disappointing.",
    "The product quality is very poor.",
    "The package arrived yesterday.",
    "The box contains a phone, a charger and a cable.",
    "The screen is beautiful, but the battery life is disappointing.",
    "I am extremely happy with my purchase.",
    "The delivery was late and the package was damaged.",
    "This product is average and nothing special."
]

data = pd.DataFrame({
    "Review": reviews
})

print("\nOriginal Review Dataset:")
display(data)

# Step 6: Sentiment Analysis using TextBlob

def analyze_textblob(text):
    sentiment = TextBlob(text).sentiment

    polarity = sentiment.polarity
    subjectivity = sentiment.subjectivity

    if polarity > 0:
        label = "Positive"
    elif polarity < 0:
        label = "Negative"
    else:
        label = "Neutral"

    return polarity, subjectivity, label


textblob_results = data["Review"].apply(analyze_textblob)

data["TextBlob_Polarity"] = [
    result[0] for result in textblob_results
]

data["TextBlob_Subjectivity"] = [
    result[1] for result in textblob_results
]

data["TextBlob_Sentiment"] = [
    result[2] for result in textblob_results
]

print("\nTextBlob Sentiment Analysis:")
display(data)

# Step 7: Sentiment Analysis using VADER

def analyze_vader(text):
    scores = vader_analyzer.polarity_scores(text)

    compound = scores["compound"]

    if compound >= 0.05:
        label = "Positive"
    elif compound <= -0.05:
        label = "Negative"
    else:
        label = "Neutral"

    return (
        scores["pos"],
        scores["neg"],
        scores["neu"],
        compound,
        label
    )


vader_results = data["Review"].apply(analyze_vader)

data["VADER_Positive"] = [
    result[0] for result in vader_results
]

data["VADER_Negative"] = [
    result[1] for result in vader_results
]

data["VADER_Neutral"] = [
    result[2] for result in vader_results
]

data["VADER_Compound"] = [
    result[3] for result in vader_results
]

data["VADER_Sentiment"] = [
    result[4] for result in vader_results
]

print("\nVADER Sentiment Analysis:")
display(data)

# Step 8: Compare TextBlob and VADER

comparison = data[
    [
        "Review",
        "TextBlob_Sentiment",
        "VADER_Sentiment"
    ]
]

print("\nTextBlob vs VADER Comparison:")
display(comparison)

# Step 9: Calculate sentiment distribution

print("\nTextBlob Sentiment Distribution:")
print(data["TextBlob_Sentiment"].value_counts())

print("\nVADER Sentiment Distribution:")
print(data["VADER_Sentiment"].value_counts())

# Step 10: Calculate agreement between both methods

agreement = (
    data["TextBlob_Sentiment"] ==
    data["VADER_Sentiment"]
)

agreement_count = agreement.sum()
total_reviews = len(data)

agreement_percentage = (
    agreement_count / total_reviews
) * 100

print("\nSentiment Agreement Analysis")
print("-" * 40)

print("Total Reviews:", total_reviews)
print("Agreed Predictions:", agreement_count)
print("Disagreed Predictions:", total_reviews - agreement_count)

print(f"Agreement Percentage: {agreement_percentage:.2f}%")

# Step 11: Analyze a new review

new_review = (
    "The camera quality is fantastic, "
    "but the battery performance is poor."
)

print("\nNew Review:")
print(new_review)

tb_result = TextBlob(new_review).sentiment
vader_result = vader_analyzer.polarity_scores(new_review)

print("\nTextBlob Results:")
print("Polarity:", tb_result.polarity)
print("Subjectivity:", tb_result.subjectivity)

print("\nVADER Results:")
print(vader_result)

# Step 12: Classify the new review

if tb_result.polarity > 0:
    tb_label = "Positive"
elif tb_result.polarity < 0:
    tb_label = "Negative"
else:
    tb_label = "Neutral"

if vader_result["compound"] >= 0.05:
    vader_label = "Positive"
elif vader_result["compound"] <= -0.05:
    vader_label = "Negative"
else:
    vader_label = "Neutral"

print("\nFinal Sentiment Classification")
print("-" * 40)

print("TextBlob:", tb_label)
print("VADER:", vader_label)

# Step 13: Visualize sentiment distribution

textblob_counts = data["TextBlob_Sentiment"].value_counts()
vader_counts = data["VADER_Sentiment"].value_counts()

sentiment_comparison = pd.DataFrame({
    "TextBlob": textblob_counts,
    "VADER": vader_counts
}).fillna(0)

sentiment_comparison.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Sentiment Distribution: TextBlob vs VADER")
plt.xlabel("Sentiment")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=0)
plt.legend()
plt.tight_layout()
plt.show()

# Step 14: Result and Conclusion

print("\n" + "=" * 60)
print("RESULT AND CONCLUSION")
print("=" * 60)

print("""
Result:
Sentiment analysis was successfully performed on the given
review dataset using TextBlob and VADER.

TextBlob polarity and subjectivity scores were calculated.
VADER positive, negative, neutral and compound scores were
also calculated for each review.

The results of both sentiment analysis techniques were
compared to observe differences in their classifications.

Conclusion:
Sentiment analysis is an important Natural Language Processing
technique used to identify opinions and emotions expressed
in text. TextBlob and VADER provide different approaches
for sentiment detection and analysis.
""")
```
