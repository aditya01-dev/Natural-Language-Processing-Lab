```python
# ============================================================
# NATURAL LANGUAGE PROCESSING LABORATORY
#
# Practical 18: Topic Modeling using Latent Semantic Analysis
# (LSA)
# Name: Aditya kumar Singh
# Course: B.Tech CSE (Artificial Intelligence)
# Subject: Natural Language Processing Laboratory
# Practical No.: 18
# CO: CO2
#
# Objective:
# To extract latent semantic structures from a text corpus
# using Latent Semantic Analysis (LSA) and Singular Value
# Decomposition (SVD).
# ============================================================

# Step 1: Install required libraries

!pip install -q scikit-learn pandas numpy matplotlib

# Step 2: Import libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
from sklearn.metrics.pairwise import cosine_similarity

print("Libraries imported successfully.")

# Step 3: Create a sample text corpus

documents = [
    # Technology documents
    "Artificial intelligence and machine learning are changing technology.",
    "Deep learning algorithms are used in artificial intelligence.",
    "Python programming is useful for software development.",
    "Computer networks connect devices through digital communication.",
    "Machine learning models analyze data and make predictions.",

    # Sports documents
    "Football players practice before the championship tournament.",
    "Cricket players improve their batting and bowling skills.",
    "Basketball teams compete in professional sports leagues.",
    "Athletes participate in international sports competitions.",
    "Football and cricket are popular games around the world.",

    # Food documents
    "Cooking delicious food requires fresh vegetables and spices.",
    "Italian pizza and pasta are popular food dishes.",
    "Healthy eating includes fruits and nutritious vegetables.",
    "Chefs prepare delicious recipes using different cooking methods.",
    "Fresh fruits and vegetables are important for a healthy diet."
]

print("\nOriginal Text Corpus:\n")

for i, document in enumerate(documents, 1):
    print(f"Document {i}: {document}")

# Step 4: Convert documents into TF-IDF vectors

vectorizer = TfidfVectorizer(stop_words="english")

tfidf_matrix = vectorizer.fit_transform(documents)

feature_names = vectorizer.get_feature_names_out()

print("\nTF-IDF Matrix Shape:", tfidf_matrix.shape)

# Step 5: Display the TF-IDF matrix

tfidf_table = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=feature_names
)

print("\nTF-IDF Representation:")
display(tfidf_table.round(3))

# Step 6: Apply Singular Value Decomposition (SVD)

lsa_model = TruncatedSVD(
    n_components=3,
    n_iter=10,
    random_state=42
)

lsa_matrix = lsa_model.fit_transform(tfidf_matrix)

print("\nLSA Model Created Successfully.")
print("LSA Matrix Shape:", lsa_matrix.shape)

# Step 7: Display latent semantic components

print("\nLatent Semantic Components:")
print("=" * 60)

for component_index, component in enumerate(
    lsa_model.components_
):
    top_indices = np.argsort(np.abs(component))[-10:][::-1]

    top_terms = [
        feature_names[index]
        for index in top_indices
    ]

    print(f"\nComponent {component_index + 1}:")
    print(", ".join(top_terms))

# Step 8: Display component weights

print("\nComponent-Term Weights:")

for component_index, component in enumerate(
    lsa_model.components_
):
    top_indices = np.argsort(np.abs(component))[-10:][::-1]

    component_table = pd.DataFrame({
        "Term": feature_names[top_indices],
        "Weight": component[top_indices]
    })

    print(f"\nComponent {component_index + 1}:")
    display(component_table.round(4))

# Step 9: Display document representations in latent space

lsa_columns = [
    f"Component {i + 1}"
    for i in range(lsa_model.n_components)
]

lsa_table = pd.DataFrame(
    lsa_matrix,
    columns=lsa_columns
)

lsa_table.index = [
    f"Document {i + 1}"
    for i in range(len(documents))
]

print("\nDocument-Latent Component Matrix:")
display(lsa_table.round(4))

# Step 10: Identify the dominant component for each document

dominant_components = (
    np.argmax(np.abs(lsa_matrix), axis=1) + 1
)

document_results = pd.DataFrame({
    "Document": documents,
    "Dominant Component": dominant_components
})

print("\nDominant Latent Component:")
display(document_results)

# Note: A dominant component is an illustrative summary of
# the largest absolute document loading. LSA components are
# not probabilistic topic memberships.

# Step 11: Display explained variance

variance = lsa_model.explained_variance_ratio_

variance_table = pd.DataFrame({
    "Component": lsa_columns,
    "Explained Variance Ratio": variance
})

print("\nExplained Variance by Component:")
display(variance_table.round(4))

print(
    "\nTotal Explained Variance:",
    round(np.sum(variance), 4)
)

# Step 12: Calculate document similarity in LSA space

similarity_matrix = cosine_similarity(lsa_matrix)

similarity_table = pd.DataFrame(
    similarity_matrix,
    index=[f"Document {i+1}" for i in range(len(documents))],
    columns=[f"Document {i+1}" for i in range(len(documents))]
)

print("\nDocument Similarity Matrix in LSA Space:")
display(similarity_table.round(3))

# Step 13: Visualize documents in two-dimensional LSA space

plt.figure(figsize=(9, 6))

plt.scatter(
    lsa_matrix[:, 0],
    lsa_matrix[:, 1]
)

for i in range(len(documents)):
    plt.annotate(
        f"D{i+1}",
        (lsa_matrix[i, 0], lsa_matrix[i, 1])
    )

plt.title("Document Visualization using LSA")
plt.xlabel("Latent Component 1")
plt.ylabel("Latent Component 2")
plt.grid(True)
plt.tight_layout()
plt.show()

# Step 14: Transform new documents

new_documents = [
    "Artificial intelligence improves machine learning.",
    "Football players prepare for the sports tournament.",
    "Fresh fruits and vegetables are used in cooking."
]

new_tfidf = vectorizer.transform(new_documents)

new_lsa = lsa_model.transform(new_tfidf)

new_lsa_table = pd.DataFrame(
    new_lsa,
    columns=lsa_columns
)

new_lsa_table.insert(
    0,
    "New Document",
    new_documents
)

print("\nNew Document LSA Representations:")
display(new_lsa_table.round(4))

# Step 15: Display model information

print("\nLSA Model Information")
print("-" * 45)

print("Number of Documents:", len(documents))
print("Vocabulary Size:", len(feature_names))
print("Number of Components:", lsa_model.n_components)
print("SVD Iterations:", lsa_model.n_iter)

# Step 16: Result and Conclusion

print("\n" + "=" * 60)
print("RESULT AND CONCLUSION")
print("=" * 60)

print("""
Result:
Latent Semantic Analysis was successfully implemented on
the given text corpus using TF-IDF vectorization and
Singular Value Decomposition.

The latent semantic components, important terms, document
representations, and explained variance were obtained.

Document similarity was also calculated in the reduced
semantic space.

Conclusion:
Latent Semantic Analysis is a technique used to discover
hidden semantic relationships between terms and documents.

It applies Singular Value Decomposition to a term-document
matrix to reduce dimensionality and identify latent
semantic structures in text data.

LSA is useful in document similarity, information retrieval,
document clustering, and topic exploration.
""")
```
