```python
from sklearn.feature_extraction.text import CountVectorizer

print("Library imported successfully.")
```

```python
documents = [
    "Natural language processing is interesting.",
    "Natural language processing is useful.",
    "Machine learning is useful for natural language processing."
]

print("Documents:")
for i, document in enumerate(documents, 1):
    print(f"Document {i}: {document}")
```

```python
vectorizer = CountVectorizer()

bow_matrix = vectorizer.fit_transform(documents)

print("Bag-of-Words matrix created successfully.")
```

```python
vocabulary = vectorizer.get_feature_names_out()

print("Vocabulary:")
print(vocabulary)
```

```python
print("Bag-of-Words Matrix:")
print(bow_matrix.toarray())
```

```python
import pandas as pd

bow_table = pd.DataFrame(
    bow_matrix.toarray(),
    columns=vocabulary
)

bow_table.index = [
    "Document 1",
    "Document 2",
    "Document 3"
]

print("Bag-of-Words Representation:")
display(bow_table)
```

```python
print("Word Counts in Each Document:\n")

for i, document in enumerate(bow_matrix.toarray(), 1):
    print(f"Document {i}:")

    for word, count in zip(vocabulary, document):
        if count > 0:
            print(f"  {word} -> {count}")

    print()
```

```python
new_document = [
    "Natural language processing is useful."
]

new_bow = vectorizer.transform(new_document)

print("New Document:")
print(new_document[0])

print("\nBoW Representation:")
print(new_bow.toarray())
```

```python
new_bow_table = pd.DataFrame(
    new_bow.toarray(),
    columns=vocabulary
)

display(new_bow_table)
```

```python
print("Vocabulary Index:")
print()

for index, word in enumerate(vocabulary):
    print(f"{index} -> {word}")
```

```python
# Result:
# The given text documents were successfully converted into numerical
# representations using the Bag-of-Words model.
#
# Conclusion:
# The Bag-of-Words model represents text using word-frequency counts.
# It converts unstructured text into numerical vectors that can be used
# as input for classical machine learning and NLP algorithms.
```
