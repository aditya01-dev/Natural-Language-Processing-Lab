```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd

print("Libraries imported successfully.")
```

```python
documents = [
    "Natural language processing is a field of artificial intelligence.",
    "Natural language processing is used to understand human language.",
    "Computer networks are used for communication between devices."
]

print("Documents:\n")

for i, document in enumerate(documents, 1):
    print(f"Document {i}: {document}")
```

```python
vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(documents)

print("TF-IDF Matrix:")
print(tfidf_matrix.toarray())
```

```python
vocabulary = vectorizer.get_feature_names_out()

print("Vocabulary:")
print(vocabulary)
```

```python
tfidf_table = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=vocabulary
)

tfidf_table.index = [
    "Document 1",
    "Document 2",
    "Document 3"
]

print("TF-IDF Vector Representation:")
display(tfidf_table.round(3))
```

```python
similarity_matrix = cosine_similarity(tfidf_matrix)

print("Cosine Similarity Matrix:")
print(similarity_matrix.round(3))
```

```python
similarity_table = pd.DataFrame(
    similarity_matrix,
    index=["Document 1", "Document 2", "Document 3"],
    columns=["Document 1", "Document 2", "Document 3"]
)

print("Cosine Similarity Table:")
display(similarity_table.round(3))
```

```python
print("Pairwise Cosine Similarity:\n")

print(
    f"Document 1 vs Document 2: "
    f"{similarity_matrix[0][1]:.4f}"
)

print(
    f"Document 1 vs Document 3: "
    f"{similarity_matrix[0][2]:.4f}"
)

print(
    f"Document 2 vs Document 3: "
    f"{similarity_matrix[1][2]:.4f}"
)
```

```python
import numpy as np

vector_1 = tfidf_matrix.toarray()[0]
vector_2 = tfidf_matrix.toarray()[1]

manual_similarity = np.dot(vector_1, vector_2) / (
    np.linalg.norm(vector_1) * np.linalg.norm(vector_2)
)

print("Manual Cosine Similarity:")
print(f"{manual_similarity:.4f}")

print("\nScikit-learn Cosine Similarity:")
print(f"{similarity_matrix[0][1]:.4f}")
```

```python
doc1 = "Natural language processing is interesting."
doc2 = "Natural language processing is useful."

test_documents = [doc1, doc2]

test_vectorizer = TfidfVectorizer()
test_vectors = test_vectorizer.fit_transform(test_documents)

test_similarity = cosine_similarity(test_vectors)

print("Document 1:")
print(doc1)

print("\nDocument 2:")
print(doc2)

print(
    f"\nCosine Similarity: "
    f"{test_similarity[0][1]:.4f}"
)
```

```python
score = similarity_matrix[0][1]

print("Interpretation of Cosine Similarity")
print("-" * 45)
print(f"Similarity Score: {score:.4f}")

if score >= 0.7:
    print("The documents are highly similar.")
elif score >= 0.4:
    print("The documents are moderately similar.")
else:
    print("The documents have low similarity.")
```

```python
# Cosine Similarity Formula:
#
#    A · B
# --------------------------------
#   ||A|| × ||B||
#
# Where:
# A = Vector representing Document 1
# B = Vector representing Document 2
#
# Cosine similarity ranges from -1 to 1 in the general vector case.
# For standard TF-IDF vectors, the values are non-negative and therefore
# normally lie between 0 and 1.
```
